# Working with CSV

Let's import a file with fuel economy data

```python
import csv
with open('mpg.csv') as csvfile:
  mpg = list(csv.DictReader(csvfile))
print(mpg)
```

Find out how many items are there in the dictionary

```python
i=0
for item in mpg[1].items():
  print(item)
  i+=1
print('no. of items:',i)
```

Find the column names with keys

```python
columns = mpg[0].keys()
print(columns)
```

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

```python
sum_cty = 0
for car in mpg:
    sum_cty += int(car['cty'])
avg_cty = sum_cty/len(mpg)
print("Average City Fuel Economy:",avg_cty)
```

Find the average hwy fuel economy across all cars

```python
sum_hwy = 0
for car in mpg:
    sum_hwy += int(car['hwy'])
avg_hwy = sum_hwy/len(mpg)
print("Average Highway Fuel Economy:",avg_hwy)
```

Use set to return the unique values for the number of cylinders the cars.

```python
cylinders = set()
for d in mpg:
  cylinders.add(d['cyl'])
print(cylinders)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```python
MpgByCyl = []
for c in cylinders:
    summpg = 0
    typecount = 0
    for d in mpg:
        if d['cyl'] == c:
            summpg += int(d['cty'])
            typecount += 1
    MpgByCyl.append((c, summpg / typecount))
MpgByCyl.sort(key=lambda x: x[1])
print(MpgByCyl)
```

Return the unique values for the class types in our dataset using set.

```python
vehicleclass = set(d['class'] for d in mpg)
print(vehicleclass)
```

Find average miles per gallon in the highway for each class of vehicles

```python
MileageByClass = []
for t in vehicleclass:
    total_hwy_mpg = 0
    count = 0
    for d in mpg:
        if d['class'] == t:
            total_hwy_mpg += int(d['hwy'])
            count += 1
    MileageByClass.append((t, total_hwy_mpg/count))
MileageByClass.sort(key=lambda x: x[1])
print(MileageByClass)
```
